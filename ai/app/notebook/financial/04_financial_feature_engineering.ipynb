# Financial Feature Engineering

Create model-ready features without storing the dataset in Git.

```python
from pathlib import Path
import pandas as pd

df = pd.read_csv(Path('../../../datasets/paysim.csv'))
df['balance_delta'] = df['newbalanceOrig'] - df['oldbalanceOrg']
df['recipient_balance_delta'] = df['newbalanceDest'] - df['oldbalanceDest']
df['is_zero_balance_origin'] = (df['oldbalanceOrg'] == 0).astype('int8')
df.head()
```
